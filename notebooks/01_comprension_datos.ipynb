# 01 · Comprensión de los datos

Exploramos los archivos de Matemáticas y Portugués del conjunto Student Performance. Cada fila representa un registro de estudiante en una asignatura. La documentación señala que 382 estudiantes aparecen en ambos archivos; por eso los analizamos por separado y no sumamos las filas como si fueran personas distintas.

**Objetivo:** conocer las fuentes, sus variables y su calidad antes de hacer comparaciones.

## 1. Librerías y lectura de los archivos

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

raiz = Path.cwd()
if not (raiz / 'datos').is_dir():
    raiz = raiz.parent
if not (raiz / 'datos').is_dir():
    raise FileNotFoundError('Abre el proyecto o la carpeta notebooks antes de ejecutar.')
resultados = raiz / 'resultados'
resultados.mkdir(exist_ok=True)
fuentes = {'matematicas': 'student-mat.csv', 'portugues': 'student-por.csv'}
datos = {nombre: pd.read_csv(raiz / 'datos' / archivo, sep=';')
         for nombre, archivo in fuentes.items()}
```

## 2. Inventario de las fuentes
Comprobamos dimensiones, celdas vacías y filas completamente duplicadas.

```python
inventario = pd.DataFrame([
    {'Asignatura': nombre.capitalize(), 'Archivo': fuentes[nombre],
     'Registros': len(df), 'Columnas': len(df.columns),
     'Celdas vacías': int(df.isna().sum().sum()), 'Filas duplicadas': int(df.duplicated().sum())}
    for nombre, df in datos.items()
])
display(inventario)
assert list(inventario['Registros']) == [395, 649]
assert inventario['Celdas vacías'].sum() == 0
```

En los archivos revisados no hay celdas vacías ni filas exactamente iguales. Esto no garantiza que todas las respuestas sean correctas: la revisión solo detecta esos dos problemas concretos.

## 3. Variables del análisis

```python
variables = pd.DataFrame([
    ('studytime', 'Tiempo de estudio semanal', 'Categoría ordenada de 1 a 4'),
    ('goout', 'Salidas con amigos', 'Categoría ordenada de 1 a 5'),
    ('Dalc', 'Consumo de alcohol entre semana', 'Categoría ordenada de 1 a 5'),
    ('Walc', 'Consumo de alcohol en fin de semana', 'Categoría ordenada de 1 a 5'),
    ('absences', 'Número de ausencias', 'Numérica discreta'),
    ('failures', 'Fracasos académicos anteriores', 'Numérica discreta'),
    ('G1', 'Nota del primer período', 'Numérica de 0 a 20'),
    ('G2', 'Nota del segundo período', 'Numérica de 0 a 20'),
    ('G3', 'Nota final', 'Numérica de 0 a 20; variable objetivo')
], columns=['Variable', 'Descripción', 'Interpretación'])
display(variables)
```
